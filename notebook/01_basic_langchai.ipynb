```python
### now lets say instead of english we want to translate to hindi
# here we can use prompt template

from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model="gemini-3.6-flash",
    model_provider="google_genai"
)

prompts = ChatPromptTemplate.from_messages([ # dyanmic prompts
    {"role": "system", "content": "You are a Language Translator and translate from English to {language}."},
    {"role": "user", "content": "{query}"},
])

prompt = prompts.format(language="Hinglish", query="I'm Sandeep and I'm a coder")

response = llm.invoke(prompt)
print(response.content[0]['text'])
```

```python
# Create an LCEL Chain
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model

from langchain_core.output_parsers import StrOutputParser

llm = init_chat_model(
    model="gemini-2.5-flash",  # or your preferred model identifier
    model_provider="google_genai"
)

prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a Language Translator and translate from English to {language}."},
    {"role": "user", "content": "{query}"},
])

output = StrOutputParser()

# custom prompt template
def transform_case(self):
    return self.upper()

#Instead of invoking the prompt and the LLM separately, chain them together using the Pipe (|) operator. This is the standard Runnable pattern in LangChain:
chain = prompts | llm | output | transform_case

# Pass the input dictionary to the chain
res = chain.invoke({"language": "Hinglish", "query": "I'm Sandeep and I'm a coder"})

print(res)
```
