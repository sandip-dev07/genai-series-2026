### Text Spliter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

texts = "Cricket in 2026 has been a major year for the sport, with international cricket bringing together \
established teams, emerging nations, and millions of fans around the world. The biggest men's \
event of the year was the ICC Men's T20 World Cup, co-hosted by India and Sri Lanka."

splitter = RecursiveCharacterTextSplitter(chunk_size = 50, chunk_overlap = 20)
text_splitter = splitter.split_text(texts)

text_splitter
```

### Document Text Loader

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("../data/2026_cricket.pdf")
documents = loader.load()

splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
doc_splitter = splitter.split_documents(documents)

doc_splitter
```
