```python
from langchain_ollama import ChatOllama

model = ChatOllama(
    model="qwen2.5-coder:3b",
)
```

```python
messages = [
    (
        "system",
        "You are a Software Developer",
    ),
    ("human", "two sum leetcode question, c++ code please"),
]

response = ""

for chunk in model.stream("Can you explain how AI works in detail?"):
    response += chunk.content
    print(chunk.content, end="", flush=True)
```
