### Store Embeddings in Vector DB

```python
from langchain_chroma import Chroma
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_google_genai import GoogleGenerativeAIEmbeddings


# 1. Load PDF
loader = PyPDFLoader("../data/2026_cricket.pdf")
documents = loader.load()


# 2. Split documents into chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

doc_splitter = splitter.split_documents(documents)


# 3. Create embedding model
embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-2-preview"
)


# 4. Create Chroma vector store
vector_store = Chroma.from_documents(
    documents=doc_splitter,
    embedding=embeddings,
    persist_directory="./vector_db"
)

print("Documents stored successfully!")
```

```python
query = "Cricket"

res = vector_store.similarity_search(query=query, k=3)

for i, doc in enumerate(res, 1):
    print(f"\n--- Result {i} ---")
    print(doc.page_content)
    print("Metadata:", doc.metadata)
```
